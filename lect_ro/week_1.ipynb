# Populația statistică și eșantionul

**Populația statistică** (din lat. *generis* — comun, generic) — totalitatea obiectelor despre care intenționăm să tragem concluzii în cadrul unei probleme concrete, în continuare **PS**.<br>

**Eșantionul** — un număr mic de obiecte din **PS**. În practică nu putem măsura întreaga **PS**.

**Exemplu:** **PS** — toți cei 10 000 de studenți ai universității, dintre care 30% sunt fumători; **eșantionul** — 500 de studenți selectați dintre acești 10 000 de studenți.

**Reprezentativitatea** — gradul de asemănare a obiectelor dintre **PS** și **eșantion**.

**Eșantionul reprezentativ** — un model cât mai asemănător cu **PS**, dar la scară redusă, în continuare **ER**.

**Exemplu**: dacă în **PS** 30% sunt fumători, iar 25% frecventează săli de fitness, atunci în componența **ER** trebuie păstrate aceste proporții.

## Metode de eșantionare reprezentativă

Tipuri de eșantioane:

Eșantioane probabilistice — la formarea acestor eșantioane presupunem că populația statistică este omogenă și că toate elementele ei sunt la fel de accesibile.

    * Eșantionul aleator simplu (simple random sample) – un set aleator de obiecte din populația statistică. Exemplu: 100 de bărbați care participă la Jocurile Olimpice

    * Eșantionul stratificat (stratified sample) – înainte de a selecta aleator obiectele din populația statistică, o împărțim în mai multe straturi (grupuri). Exemplu: bărbați de 18-25 ani, 26-31, 32-36 ș.a.m.d. Apoi, din aceste grupuri, alegem aleator câte N persoane.

    * Eșantionul pe grupuri / cluster (cluster sample) – mai întâi împărțim populația statistică în clustere, considerând că acestea sunt asemănătoare între ele. Exemplu: înălțimea locuitorilor din Sankt Petersburg. Îi împărțim pe raioane (Admiralteiski, Vasileostrovski etc.), apoi selectăm aleator persoane din câteva raioane alese la întâmplare pentru cercetare.

Eșantioane neprobabilistice — selecția într-un astfel de eșantion nu se face după principiul aleatoriului, ci după criterii subiective – accesibilitatea obiectelor, tipicitatea sau reprezentarea egală. Astfel de eșantioane se întâlnesc des în cercetările sociologice, însă datele obținute pe baza lor au o fiabilitate mai mică și e mai bine să le evităm.

    * Metoda bulgărelui de zăpadă — fiecărui respondent, începând cu primul, i se cer contactele prietenilor, colegilor, cunoscuților care ar corespunde criteriilor de selecție și ar putea participa la cercetare. Problema principală a unui asemenea eșantion — nu este atins un grup aleator de persoane, ci persoane legate prin interese comune, hobby-uri etc.

    * Eșantionul spontan — se chestionează respondenții cei mai accesibili. Mărimea și componența eșantioanelor spontane nu sunt cunoscute dinainte și sunt determinate de un singur parametru – activitatea respondenților. Exemplu: un sondaj realizat într-un ziar sau într-o revistă, majoritatea sondajelor pe internet.

    * Eșantionul cazurilor tipice — se selectează unitățile populației statistice care au o valoare tipică a caracteristicii (adesea valoarea medie). Aici apare problema alegerii caracteristicii și a determinării valorii ei tipice.

# Tipuri de variabile

![](../to_add/col_nomin.png)

Toate variabilele pot fi împărțite în două grupuri principale:

### 1. Cantitative (numerical) — valori numerice

Variabile ale căror valori sunt exprimate prin numere și au sens cantitativ.

&nbsp;&nbsp;&nbsp;&nbsp;**a) Continue (continuous)** — pot lua orice valoare într-un anumit interval de numere.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**Exemple:** înălțimea, greutatea, temperatura, timpul.
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;De exemplu: înălțimea poate fi `175`, `175.5`, `175.52` cm.

&nbsp;&nbsp;&nbsp;&nbsp;**b) Discrete (discrete)** — iau valori separate, de obicei numere întregi; astfel de valori pot fi numărate.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**Exemple:** numărul de copii, de cărți, de publicații, de goluri marcate.
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;De exemplu: `1`, `2`, `3`, `4` copii. `3.5` copii nu pot exista.

### 2. Calitative (categorical) — categorii sau grupuri

Variabile care împart obiectele în anumite categorii. Numerele din astfel de variabile pot fi folosite doar ca **coduri (etichete)** și nu au sens cantitativ.

&nbsp;&nbsp;&nbsp;&nbsp;**a) Nominale (nominal)** — împărțirea obiectelor în grupuri între care **nu există o ordine anume**.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**Exemple:**
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;`1 — femeie`, `2 — bărbat`;
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;grupa sanguină: `A`, `B`, `AB`, `O`;
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;profesia: `profesor`, `medic`, `inginer`;
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;culoarea ochilor: `căprui`, `albaștri`, `verzi`.

&nbsp;&nbsp;&nbsp;&nbsp;**b) Ordinale (ordinal)** — categorii între care există **o ordine sau o ierarhie**. Astfel de variabile pot fi comparate între ele, dar diferența dintre ranguri vecine nu este neapărat aceeași.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**Exemple:**
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;locurile într-o cursă: `1`, `2`, `3`;
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;nivelul de educație: `primar` → `mediu` → `superior`;
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;satisfacția: `nesatisfăcător` → `satisfăcător` → `bine` → `excelent`;
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;clasa de servire: `economy` → `business` → `first class`.

### Esențialul

Variabilele **cantitative** — valori care pot fi măsurate sau numărate.

Variabilele **calitative** — categorii:

- **nominale** — categorii **fără ordine**;
- **ordinale** — categorii **cu ordine**.

De exemplu, `bărbat / femeie` — variabilă nominală, iar `locul 1 / locul 2 / locul 3` — variabilă ordinală.

# Măsuri ale tendinței centrale

<center><img src="../slides/week_1/week_1-06.png" title="slide 6" width="50%" height="50%"/></center>

`Măsura tendinței centrale` — un număr care servește la descrierea unei mulțimi de valori printr-un singur număr (pentru concizie).
**De exemplu**, în loc să enumerăm salariile tuturor angajaților unei organizații, vorbim despre salariul mediu. Există multe măsuri ale tendinței centrale; alegerea finală a măsurii rămâne întotdeauna la latitudinea cercetătorului.

<center><img src="../slides/week_1/week_1-07.png" title="slide 7" width="50%" height="50%"/></center>

`Modul` (tipicitatea) - valoarea care apare cel mai frecvent. Uneori într-o mulțime există mai mult de un mod.
**Grafic**: dot plot.
**De exemplu**: $[6, 2, 6, 6, 8, 9, 9, 9, 0]$, $M_{o} = 6 , 9$, mulțimea este multimodală.

```python
from scipy import stats
import numpy as np
a = np.array([185,175,170,169,171,172,175,157,170,172,167,173,168,167,166,
167,169,172,177,178,165,161,179,159,164,178,172,170,173,171])

m = stats.mode(a)
print(m)
```

`Mediana` - valoarea care împarte observațiile __ordonate crescător (descrescător)__ în două jumătăți. Dacă mulțimea conține un număr par de elemente, se ia media celor două elemente din mijloc ale mulțimii ordonate.
**Exemplu**: $[1, 2, 50, 100, 1000, 2000000]$, $M_{e} =\frac{(50+100)}{2} = 75$.

```python
import numpy as np

a = np.array([185,175,170,169,171,172,175,157,170,172,167,173,168,167,166,
 167,169,172,177,178,165,161,179,159,164,178,172,170,173,171])
m = np.median(a)
print(m)
```

<center><img src="../slides/week_1/week_1-10.png" title="slide 10" width="50%" height="50%"/></center>

`Media` (*mean*, media aritmetică) - numărul egal cu suma tuturor numerelor mulțimii, împărțită la numărul lor. Cazuri particulare ale mediei aritmetice sunt media (populației statistice, notată cu litera greacă $\mu$ miu mic) și media de eșantion (a eșantionului, notată $\overline X$).

## Exemple de utilizare

- Dacă distribuția este simetrică, unimodală și nu are valori aberante vizibile: se poate folosi oricare dintre măsurile tendinței centrale — media, modul și mediana vor da aproximativ aceeași valoare.
- Dacă distribuția are o asimetrie pronunțată (înclinată spre stânga/dreapta), valori aberante vizibile sau mai multe moduri: utilizarea mediei nu este relevantă; e mai bine să folosim modul sau mediana pentru a caracteriza datele din punctul de vedere al intensității unei anumite caracteristici cantitative.

<img src="../to_add/diff.png" width="600" height="400">

```python
from scipy import stats
import numpy as np

a = np.array([185, 175, 170, 169, 171, 175, 157, 172, 170, 172, 167, 173, 168, 167, 166,
              167, 169, 172, 177, 178, 165, 161, 179, 159, 164, 178, 172, 170, 173, 171])
# cum se caută modul
m = stats.mode(a)
print(m)
# cum se caută mediana
m = np.median(a)
print(m)
# cum se caută media
m = np.mean(a)
print(m)
```

`Amplitudinea` (*range*, împrăștierea) eșantionului măsoară distanța în limitele căreia variază valorile și se notează cu litera $R = X_{(n)} - X_{(1)}$, diferența dintre valoarea maximă și cea minimă a șirului.

`Speranța matematică` - adesea numită pur și simplu media.

Pentru a înțelege ce abatere este tipică pentru o mulțime, adunăm toate abaterile și împărțim la numărul de elemente. Însă vom obține 0, pentru că unele abateri sunt pozitive, iar altele negative. Trebuie să scăpăm de semn; sunt 2 variante: fie luăm modulul abaterilor, fie le ridicăm la pătrat (varianta aleasă mai des). Dacă găsim media pătratelor abaterilor, obținem dispersia.

`Dispersia` (*variance*) - măsura împrăștierii valorilor unei variabile aleatoare față de speranța ei matematică (valoarea medie).
- O dispersie mică înseamnă că valorile sunt grupate aproape una de alta.
- O dispersie mare indică o împrăștiere puternică a valorilor.

**Dispersia populației statistice**:
$$D = \frac{\sum_{i=1}^{n} (x_{i} - M_{x})^2}{n}$$
**Dispersia eșantionului**:
$$D = \frac{\sum_{i=1}^{n} (x_{i} - M_{x})^2}{n-1}$$

- scăderea lui -1 este legată de noțiunea de grade de libertate, pe care o vom analiza mai târziu

Pătratul din această formulă face dispersia foarte incomodă pentru evaluarea diversității elementelor, din cauza creșterii dimensiunii. De aceea, pentru comoditate, extragem rădăcina din dispersie, obținând în final abaterea medie pătratică.

`Abaterea medie pătratică` - cel mai răspândit indicator al împrăștierii valorilor unei variabile aleatoare față de speranța ei matematică (valoarea medie).

$$ \sigma = \sqrt{D}$$

- **Abaterea medie pătratică** - rădăcina din dispersia populației statistice.
- **Abaterea standard** - rădăcina din dispersia eșantionului.

Dacă am fi calculat dispersia folosind modulul în loc de ridicarea la pătrat, am fi obținut abaterea liniară medie ([dezavantaje](http://univer-nn.ru/statistika/srednee-linejnoe-otklonenie/)).
Dispersia și abaterea medie pătratică sunt la fel de sensibile la valorile aberante ca și media aritmetică.


Ideea este că din valoarea pătratică se poate lua derivata (se numește netezimea funcției, iar acest lucru este foarte important pentru metodele matematice). Iar funcția modul are o „ruptură” în punctul 0, ceea ce este foarte rău. De aceea este necesară suma pătratelor erorilor, nu suma modulelor erorilor.

<center><img src="../to_add/disp_calc.png" title="slide 14" width="50%" height="50%"/></center>

## Proprietățile dispersiei și ale abaterii standard

<center>
    <img src="../slides/week_1/week_1-16.png" title="slide 16" width="50%" height="50%"/></center>

$$ D_{x+c} = D_x $$
$$ sd_{x+c} = sd_{x} $$

$$ D_{x*c} = D_x*c^2 $$
$$ sd_{x*c} = sd_{x}*C $$

$$
D[cx] = M[(cx)^2] - (M[cx])^2 = c^2 M[x^2] - c^2(M[x])^2 = c^2 D[x]
$$

$$
sd_{c*x} = \sqrt{D[c*x]} = \sqrt{c^2 D[x]} = |c| \cdot sd_x
$$

## Exemple

```python
'''
Calculul amplitudinii, dispersiei și abaterii standard
'''
A = np.array([185, 175, 170, 169, 171, 175, 157, 172, 170, 172, 167, 173, 168, 167, 166,
              167, 169, 172, 177, 178, 165, 161, 179, 159, 164, 178, 172, 170, 173, 171])

print("Amplitudine {}, min {}, max {}".format(
    np.ptp(A), # "peak to peak"
    np.nanmin(A),
    np.nanmax(A)
))

print("Dispersie {}".format(
    np.var(A, ddof=1) # ddof - delta gradelor de libertate (N-ddof)
))

print("Abatere standard {}".format(
    np.std(A, ddof=1) # ddof - delta gradelor de libertate (N-ddof)
))
```

```python
'''
Exemplu de calcul al speranței matematice, al abaterilor de la medie, al dispersiei și al abaterii medii pătratice
'''
A = [1,5,2,7,1,9,3,8,5,9]
B = []

Mx = sum(A)/(len(A))     # speranța matematică

for el in A:
    B.append((el-Mx)**2) # abaterile de la medie

D = sum(B)/(len(B)-1)    # dispersia eșantionului

Sd = D ** .5             # abaterea medie pătratică

print(Mx, B, D, Sd, sep="\n")
```

# Distribuția normală

<center><img src="../slides/week_1/week_1-20.png" title="slide 20" width="50%" height="50%"/></center>

`Distribuția normală` (distribuția Gauss sau Gauss - Laplace) -  distribuție de probabilitate care, în cazul unidimensional, este dată de funcția de densitate de probabilitate ce coincide cu funcția Gauss:
$$f(x) = \frac{1} {\sigma\sqrt{2 \pi}} exp{( -\frac{(x- \mu )^2} {2 \sigma ^2} )},$$
unde $\mu$ - speranța matematică (media), mediana și modul distribuției, iar $\sigma$ - abaterea medie pătratică ($\sigma^{2}$ — dispersia) distribuției.

- Distribuție unimodală - distribuție care are un singur mod (adică un singur „vârf”).
- Este simetrică.
- Abaterile observațiilor de la medie se supun unei anumite legi probabilistice.

## Scurt istoric

Dacă ne oprim puțin mai în detaliu, distribuția normală apare ca rezultat al acțiunii unei mulțimi de factori, contribuția fiecăruia fiind foarte mică.

Pentru a ușura această înțelegere, în 1873 Francis Galton a construit un dispozitiv numit ulterior Planșa lui Galton (sau quincunx). Ideea e simplă: de sus, prin mijloc, sunt lansate bile care, trecând prin mai multe niveluri (de exemplu, 10), la fiecare nivel se lovesc de un obstacol și, la fiecare ciocnire, sar fie spre stânga, fie spre dreapta (cu probabilitate egală).

<center><img src="../imgs/galton_board.gif" title="Galton board" width="30%" height="30%"/></center>

Încă un [exemplu 1](https://www.youtube.com/watch?v=7NUGpzspLD4) (1 min 40 sec) și [exemplul 2](https://www.youtube.com/watch?v=GANHXlMpVMc) (2 min 45 sec).

```python
'''
Simularea planșei lui Galton
'''
D = dict()
N = 10000  # numărul de bile
level = 20 # numărul de niveluri
for _ in range(N):
    index = 0
    for _ in range(level):
        index += np.random.choice([-1, 1])
    D.setdefault(index, 0)
    D[index] += 1

sns.barplot(x=list(D.keys()), y=list(D.values()));
```

## Standardizarea Z

<center><img src="../slides/week_1/week_1-21.png" title="slide 21" width="50%" height="50%"/></center>

`Distribuția normală standard` - distribuția normală cu speranța matematică $\mu=0$ și abaterea standard $\sigma = 1$.

`Standardizarea` sau `standardizarea z` - procedura de transformare a distribuției normale $X \backsim N(\mu , \sigma)$ în distribuția standard $Z \backsim N(0 , 1)$ cu scala Z (*Z-scores*).

$$ Z_{i}=\frac{x_{i} - \bar{X}}{sd} $$

Uneori este necesar să calculăm valoarea Z doar pentru o singură observație, pentru a afla cât de mult se abate ea de la medie în unități de abatere standard.

# Regula 2 și 3 sigma

<center><img src="../imgs/sigmas.svg" title="2 3 sigmas" width="50%" height="50%"/></center>

Regula care afirmă că probabilitatea ca o variabilă aleatoare să se abată de la speranța sa matematică cu mai mult de trei abateri medii pătratice este practic egală cu zero. Regula este valabilă doar pentru variabilele aleatoare distribuite după legea normală.

## Exemplu

Materiale suplimentare pentru găsirea procentului de observații din intervalul care ne interesează:

- [tabelul z](https://web.archive.org/web/20180729102938/http://users.stat.ufl.edu/~athienit/Tables/Ztable.pdf): procentul observațiilor care nu depășesc valoarea z indicată.
- [tabelul z](http://www.normaltable.com/ztable-righttailed.html): procentul observațiilor care depășesc valoarea z indicată.
- [serviciu](https://gallery.shinyapps.io/dist_calc/) interactiv (default: $M=0$, $sd=1$)

```python
'''Exemplul 1
Se consideră că valoarea IQ (nivelul de inteligență) la oameni are o distribuție normală
cu media 100 și abaterea standard 15 (M = 100, sd = 15).
Aproximativ ce procent de oameni au IQ > 125?
'''
M, sd, iq = 100, 15, 125

print("Doar {}% dintre oameni au IQ > {}".format(
    stats.norm(M, sd).sf(iq) * 100,
    iq
    ))

# SF (Survival function) = (1 - cdf)
# CDF (Cumulative distribution function)
```

**Rezolvarea în text a problemei**:
1. Transformăm $IQ=125$ pe scala z, pentru a obține poziția lui $125$ pe scala z:
$$z_{125}= \frac{125-100}{15}=\frac{25}{15}=1.67$$

2. Ne uităm în tabelul valorilor care depășesc $z$. Pe verticală găsim $1.6$, iar pe orizontală $0.07$ și la intersecție obținem procentul $0.0475$.

3. Găsim procentul $0.0475*100=4.75$.

```python
'''Exemplul 2
Se consideră că valoarea IQ (nivelul de inteligență) la oameni are o distribuție normală
cu media 100 și abaterea standard 15 (M = 100, sd = 15).
Aproximativ ce procent de oameni au IQ în intervalul de la 70 la 112?
'''
M, sd, iq_low, iq_high = 100, 15, 70, 112

print("În intervalul [70,112] se află aproximativ {} % din toate valorile".format(
    (stats.norm.cdf(12/15) - stats.norm.cdf(-30/15)) * 100
))
```

**Rezolvarea cu ajutorul tabelului z**:
1. Transformăm $IQ=70$ și $IQ=112$ pe scala z, pentru a obține pozițiile acestor elemente pe scala z:
$$z_{70}= \frac{70-100}{15}=\frac{-30}{15}=-2$$
$$z_{112}= \frac{112-100}{15}=\frac{12}{15}=0.8$$

2. Ne uităm mai întâi în tabelul cu $z$ negativ pentru $z_{70}=-2$. Pe verticală $-2$, pe orizontală $0.00$. Rezultatul $0.0228$.
3. Ne uităm în tabelul cu $z$ pozitiv pentru $z_{112}=0.8$. Pe verticală $0.8$, pe orizontală $0.00$. Rezultatul $0.2119$.
4. În intervalul $[70, 112]$ se află aproximativ $100 \% - 2.28 \% - 21.19 \% = 76.53 \%$
    - Numărul de oameni cu IQ sub 70: $z_{<70}: 2.28 \%$
    - Numărul de oameni cu IQ peste 112: $z_{>112}: 21.19 \%$

**Rezolvarea folosind regula 2 și 3 sigma**:
1. Folosind [regula](#Regula-2-și-3-sigma) $2\sigma$ și $3\sigma$ estimăm:
    - Deoarece distribuția este normală, știm că în intervalul $[-2 \sigma , 2 \sigma]$ se află aproximativ $95.44 \%$ din valori. Spre stânga ne deplasăm cu $2 \sigma$ față de $0$, deci ne interesează doar $ \frac {95.44}{2} = 47.72 \%$ din valori.
    - În intervalul $[-1 \sigma, 1 \sigma]$ se află aproximativ $68.26 \%$ din toate valorile. Spre dreapta față de $0$ cu $1 \sigma$ înseamnă $ \frac {68.26}{2} = 34.13 \%$ din valori. Da, spre dreapta avem de fapt $0.8 \sigma$, dar rotunjim la unu.
2. Puțin mai puțin de $(47.72 \% + 34.13 \%) = 81.85 \% $ din valorile IQ se află în intervalul $[70, 112]$.

`TLC` afirmă că mulțimea mediilor eșantioanelor extrase din PS (**PS nu trebuie neapărat să aibă o distribuție normală**) va avea o distribuție aproximativ normală. Mai mult, media acestei distribuții va fi apropiată de media populației statistice, iar abaterea standard a acestei distribuții se va numi `eroarea standard a mediei` ($se$).

`Eroarea standard a mediei` - este abaterea medie pătratică a distribuției mediilor de eșantion!

**Forța și sensul TLC**: dacă asupra rezultatului acționează o mulțime de factori independenți și mărunți, atunci mărimea rezultantă va avea o distribuție apropiată de cea normală.
**A 2-a esență profundă**: la repetarea de multe ori a experimentelor, frecvența relativă de apariție a unei variabile aleatoare devine aproximativ egală cu probabilitatea ei.

Cunoscând abaterea standard a PS și mărimea eșantionului, putem calcula eroarea standard a mediei.

$$ se = \frac{\sigma}{\sqrt{N}} $$

unde $N$ - mărimea eșantionului.

Dacă mărimea eșantionului este suficient de mare (experimental $N>30$) și el este reprezentativ (foarte asemănător cu PS), atunci în locul abaterii standard a PS putem lua abaterea standard a eșantionului.

$$ se = \frac{sd}{\sqrt{N}} $$


https://numiqo.com/lab/central-limit-theorem

## O observație importantă despre TLC (numărul 2).
Probabil cel mai dificil moment: cum am luat și am înlocuit, pur și simplu, abaterea standard a PS cu cea de eșantion. Și ce dacă avem un eșantion cu peste 30 de observații, ce fel de număr magic este acesta?

Așa este, nu se întâmplă nicio magie. Și foarte curând vom lămuri definitiv acest lucru — imediat ce vom parcurge tema distribuției t în modulul al doilea. Acolo am descris în detaliu cum trebuie să calculăm eroarea standard a mediei dacă nu cunoaștem abaterea standard din populația statistică.

**Spoiler**: a lua pur și simplu abaterea standard de eșantion și a o substitui este o simplificare destul de grosieră, care nu se folosește în practică. Cum trebuie procedat, vedeți cât mai repede în modulul al doilea; în acest [pas](https://stepik.org/lesson/8081/step/7?unit=1360) al modulului al doilea s-a făcut un rezumat detaliat.

## Exemple

```python
# valorile zarului
dice = [1, 2, 3, 4, 5, 6]
# numărul de aruncări ale zarului
count = 6
# mărimea populației statistice
sp_size = 10000
# sp - Statistical population - populația statistică
sp = pd.Series(dtype=np.int64, index=range(sp_size))
for i in range(sp_size):
    value = 0
    for _ in range(count):
        value += np.random.choice(dice)
    sp[i] = value

sp.plot.hist(bins=28)
```

```python
# numărul de eșantioane
samples_count = 10
# mărimea eșantionului
sample_size = 200
samples = pd.DataFrame([
    [np.random.choice(sp) for _ in range(sample_size)] for __ in range(samples_count)
]).T

samples.hist(figsize=(16, 10), sharex=0)
plt.subplots_adjust(hspace = 0.6)
```

```python
means = samples.mean()
print("comparăm media PS {} și media mediilor eșantioanelor {}".format(
    sp.mean(),
    means.mean()
))
print("diferența: {}, eroarea standard a mediei: {}".format(
    abs(means.mean() - sp.mean()),
    means.std()
))
```

```python
# luăm un eșantion arbitrar
sample = samples[0]
print("sample mean: {}".format(sample.mean()))
print("sample se: {}".format( sample.std() / math.sqrt(sample.size)))
```

 # Intervale de încredere pentru medie

<center><img src="../slides/week_1/week_1-27.png" title="slide 27" width="50%" height="50%"/></center>

Articolul [Capul profesorului Bumbledorf](http://thinkcognitive.org/ru/blog/golova-professora-bambldorfa) despre intervalele de încredere.

Dacă avem o PS și un anumit eșantion din ea, atunci **nu putem cunoaște exact media PS** știind doar media eșantionului.
Totuși, putem spune, cu un anumit procent de siguranță, în ce interval se află media PS.
Evident, pentru noi este mai bine ca acest interval să fie cât mai mic, dar cum facem asta?

Știm:
- media mediilor eșantioanelor tinde spre media PS;
- eroarea standard a mediei descrie abaterea standard a distribuției mediilor eșantioanelor $se=\frac{sd_{x}}{\sqrt{n}}$.

Dacă luăm un eșantion aleator $X$ și găsim media lui $\bar{X}$, calculând și eroarea standard $se$, atunci putem calcula intervalul de încredere $[\bar{X} - 1.96*se; \bar{X} + 1.96*se]$ care descrie media PS într-un anumit interval cu **95% încredere**.

<center><img src="../imgs/norm_1_96.png" width=200></center>

Misteriosul număr **1,96** este numărul de sigma $\sigma$ din distribuția normală necesar pentru a acoperi **95%** din valorile acestei distribuții.

**Exemplul de pe slide**:

Media $\bar{X} = 100$, abaterea standard $sd=4$, iar la experiment au participat $n=64$ persoane.
Calculăm eroarea standard a mediei:
$$se=\frac{sd_{x}}{\sqrt{n}}=\frac{4}{\sqrt{64}}=\frac{4}{8}=0.5$$
Acum calculăm intervalul de încredere de 95%:
$$[\bar{X}-1.96*se, \bar{X}+1.96*se] = [100-1.96*0.5, 100+1.96*0.5]=[99.02, 100.98]$$
Iar pentru a fi siguri în proporție de 99%, lărgim intervalul de încredere:
$$[\bar{X}-2.58*se, \bar{X}+2.58*se]=[98.71, 101.29]$$

**Dacă am calculat intervalul de încredere de 95% pentru medie, aceasta înseamnă**:

 - <font color='crimson'>Media din PS aparține cu siguranță intervalului de încredere calculat.</font> - nu e sigur, deoarece calculăm intervalul de încredere pentru un eșantion din PS; putem fi siguri doar în proporție de 95% că media PS cade în acest interval.
 - <font color='green'>Putem fi siguri în proporție de 95% că media din PS aparține intervalului de încredere calculat.</font> - exact ceea ce am spus mai sus.
 - <font color='green'>Dacă repetăm experimentul de multe ori și pentru fiecare eșantion calculăm propriul interval de încredere, atunci în 95% din cazuri media adevărată se va afla în interiorul intervalului de încredere.</font> - corect, pentru că, privind din perspectiva eșantioanelor, pentru 95% dintre eșantioane media PS va fi inclusă în intervalele de încredere corespunzătoare, conform regulii $1.96 \sigma$.
 - <font color='crimson'>Media din populația statistică depășește cu siguranță limita inferioară a intervalului de încredere de 95%.</font> - fără comentarii.
 - <font color='crimson'>Dacă repetăm experimentul de multe ori, atunci 95% dintre mediile de eșantion vor aparține intervalului de încredere calculat de noi.</font> - un interval de încredere separat se calculează pentru fiecare eșantion în parte.

 __Dacă din lecție ați înțeles diferența dintre media PS și media eșantionului, precum și faptul că intervalul de încredere se construiește pentru eșantion, nu pentru PS, atunci răspunsurile din test se determină ușor.__
  

**Problemă**:
Dacă în exemplul nostru am dubla volumul eșantionului (cu condiția ca abaterea standard să rămână neschimbată), atunci intervalul de încredere de 95% ...
- <font color='crimson'>ar deveni mai larg</font>
- <font color='crimson'>sunt posibile ambele variante</font>
- <font color='green'>ar deveni mai îngust</font> - Cu cât eșantionul este mai mare (desigur, reprezentativ), cu atât mai mult **avem încredere** în datele noastre, prin urmare nu avem nevoie de un interval mare. Datele devin tot mai concentrate în jurul unor valori apropiate.

```python
'''Calculul lui 1.96 cu ajutorul scipy'''
p = 0.95 # 95%
# deoarece avem un interval bilateral, facem calculul
alpha = (1-p)/2
# isf - Inverse survival function (inverse of sf)
print("{} sigma".format(
    stats.norm().isf(alpha)
))
```

```python
'''Problemă
Calculați intervalul de încredere de 99% pentru următorul exemplu:
media (X cu bară) = 10, abaterea standard (sd) = 5, mărimea eșantionului (n) = 100
'''
p = 0.99
X_ = 10
sd = 5
n = 100

se = sd/n**0.5 # rădăcina din n
alpha = (1-p)/2
sigma = stats.norm().isf(alpha)

confidence_interval = X_ - sigma*se, X_ + sigma*se

print("[{:.2f};{:.2f}]".format(
    confidence_interval[0],
    confidence_interval[1]
))
```

# Ideea inferenței statistice, nivelul de semnificație p

Să analizăm un exemplu. Pentru vindecare sunt necesare în medie $M=20$ de zile. Însă noi am dezvoltat un preparat și am decis să aflăm dacă acest termen poate fi scurtat. Am recrutat un eșantion de $N=64$ de pacienți și am testat pe ei noua metodă de tratament. S-a dovedit că termenul mediu de vindecare s-a redus la $\bar{X} = 18.5$ zile. Cu abaterea medie pătratică, sau mai exact abaterea standard (fiindcă avem un eșantion, nu PS) $sd=4$.

**Ce concluzie putem trage pe baza acestor date?**
- pe de o parte, am scurtat într-adevăr termenul de vindecare;
- pe de altă parte, un astfel de rezultat putea fi obținut absolut întâmplător.

Să introducem două noțiuni foarte importante. În cercetarea noastră vor concura două ipoteze.
- ipoteza nulă $H_{0}$ va presupune că noul preparat nu are niciun efect și că media PS a pacienților care folosesc noul nostru preparat $M_{NP}=20$.
- ipoteza alternativă $H_{1}$ afirmă că preparatul nostru influențează viteza de vindecare și că media vitezei de vindecare a PS a celor care folosesc noul preparat $M_{NP}\neq20$.

Să raționăm astfel: presupunem că ipoteza nulă $H_{0}$ este adevărată; atunci, conform TLC, dacă am repeta de multe ori cercetarea, mediile de eșantion s-ar distribui normal în jurul mediei PS ($M=20$) și cu abaterea standard a distribuției noastre, adică eroarea standard a mediei $se = \frac{sd}{\sqrt{N}}=\frac{4}{\sqrt{64}}=0.5$.

Acum să răspundem la următoarea întrebare: **cât de mult s-a abătut media noastră de eșantion de la media presupusă a PS, în unități de abatere standard?**
Efectuăm transformarea z. Din media noastră de eșantion scădem media PS și împărțim la abaterea standard a distribuției noastre (distribuția mediilor de eșantion presupuse), în cazul nostru eroarea standard a mediei.
$$z=\frac{\bar{X}-M}{se}=\frac{18.5-20}{0.5}=-3$$
Aceasta înseamnă că, dacă în PS media ar fi fost cu adevărat $20$, atunci media noastră de eșantion s-ar fi abătut de la media PS cu $-3\sigma$ spre stânga.

În sfârșit, să folosim proprietățile distribuției normale pentru a calcula probabilitatea unei astfel de abateri, sau a uneia și mai pronunțate, de la medie. Vom folosi [serviciul](https://gallery.shinyapps.io/dist_calc/) pentru calcul și pentru a economisi timp.
- folosim distribuția normală;
- media egală cu 0;
- abaterea standard 1;
- both area: -3 și 3.

Probabilitatea de a ne abate de la medie cu mai mult de 3 sigma în ambele direcții:
$$P({X < -3} or {X > 3}) = 0.0027$$

**Ideea de bază a inferenței statistice**: mai întâi admitem că ipoteza nulă este adevărată (adică în PS nu există nicio diferență și nicio legătură). După aceea putem calcula probabilitatea de a fi obținut astfel de diferențe, sau unele și mai pronunțate, absolut întâmplător. Această probabilitate se numește `nivelul de semnificație p`. Și cu ajutorul acestui indicator aflăm care ipoteză este mai plauzibilă.
- Se consideră că dacă $p<0.05$ - putem accepta cu încredere ipoteza alternativă.
- Iar dacă $p>0.05$ - se consideră că nu avem suficiente motive pentru a respinge ipoteza nulă.

**PS**: Utilizarea valorilor p pentru verificarea ipotezelor nule în lucrările din medicină și științele naturii este criticată de mulți specialiști. Se remarcă faptul că utilizarea lor duce adesea la [erori de tipul I](https://ro.wikipedia.org/wiki/Erori_de_tip_I_%C8%99i_de_tip_II) (false positive)[7]. În particular, revista Basic and Applied Social Psychology (BASP) a interzis în 2015 publicarea articolelor în care se folosesc valori p. Editorii revistei au explicat aceasta prin faptul că nu este foarte greu să faci un studiu în care se obține p < 0,05, iar astfel de valori p mici devin prea des o justificare pentru cercetări de calitate slabă.

https://gallery.shinyapps.io/dist_calc/

Informații suplimentare despre nivelul de semnificație p pe [habr](https://habr.com/ru/company/stepic/blog/250527/) (în limba rusă).

`Eroarea statistică de tipul I` — situația în care este respinsă o ipoteză nulă corectă (engl. type I errors, α errors, false positive, respingere eronată).
`Eroarea statistică de tipul II` — situația în care este acceptată o ipoteză nulă incorectă (engl. type II errors, β errors, false negative, acceptare eronată).

**Să exersăm**:
Condițiile inițiale despre preparate din capitolul curent:
- Pentru vindecare sunt necesare în medie $M=20$ de zile;
- Eșantion de $N = 64$ de pacienți;
- Noul termen de vindecare $\bar{X} = 18.5$;
- Cu abaterea standard $sd = 4$.
- $se = \frac{sd}{\sqrt{N}}=\frac{4}{\sqrt{64}}=0.5$.
- $X_{min} = x - 1.96*se = 18.5 - (1.96 * 0.5) = 17.52$
- $X_{max} = x + 1.96*se = 18.5 + (1.96 * 0.5) = 19.48$

Utilizarea intervalelor de încredere este privită adesea ca o metodă alternativă de verificare a ipotezelor. În cazul nostru, dacă valoarea 20 (media presupusă din populația statistică) nu aparține intervalului de încredere de 95% calculat pe baza datelor de eșantion, avem suficiente motive să respingem ipoteza nulă. Verificați dacă rezultatele celor două abordări concordă: calculați intervalul de încredere de 95% pentru medie, pe exemplul cu testarea noului preparat.

- Intervalul de încredere cu fiabilitatea de 95%: $[17.52, 19.48]$, prin urmare 20 nu intră în acest interval.
- Ne întoarcem la condiția problemei - dacă valoarea 20 nu aparține IÎ de 95% al eșantionului, avem suficiente motive să respingem ipoteza nulă.

```python
'''
În medie, cursanții cursului de introducere în statistică obțin 115 puncte; însă,
în 2015, punctajul mediu a 144 de participanți aleși aleator a fost 118, cu abaterea standard 9.
Calculați nivelul de semnificație p pentru verificarea ipotezei nule conform căreia media punctajelor în 2015 este 115.
'''
def p_value(z_stat, alternative = 'two-sided'):

    if alternative == 'two-sided':
        return 2 * (1 - stats.norm.cdf(np.abs(z_stat)))

    if alternative == 'less':
        return stats.norm.cdf(z_stat)

    if alternative == 'greater':
        return 1 - stats.norm.cdf(z_stat)



x_0 = 115
x_1 = 118
sd = 9
n = 144

se = sd / n ** 0.5
z = (x_1 - x_0) / se

print(p_value(z, alternative='greater'))
```
